### Categorical Scatter Plots

#### Stripplot

- Used to display the distribution of a categorical variable.
- Each point represents an observation.

```python
tips = sns.load_dataset('tips')
tips.sample(1)
```

```python
# plotting stripplot using axis level fucn
sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
# plotting stripplot using figure level fucn
sns.catplot(data=tips, x='day', y='total_bill', kind='strip')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='strip', hue='sex')
```

```python
# jitter parameter
sns.catplot(data=tips, x='day', y='total_bill', kind='strip', hue='sex', jitter=0.2)
```

#### Swarmplot

- Similar to stripplot but points are adjusted to avoid overlap.
- Gives a better sense of the distribution of data points.

```python
# plotting swarmplot -> axis level
sns.swarmplot(data=tips, x='day', y='total_bill', hue='day', palette='deep')
```

```python
# plotting swarmplot -> figure level
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='sex')
```

Swarmplot is generally good for smaller datasets.

### Categorical Distribution Plots

#### Boxplot

- A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”).
- It can tell you about your outliers and what their values are.
- Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped, and if and how your data is skewed.

```python
# plotting box plot -> axis level
sns.boxplot(data=tips, x='day', y='total_bill')
```

```python
# using catplot
sns.catplot(data=tips, x='day', y='total_bill', kind='box')
```

```python
# hue
sns.catplot(data=tips, x='day', y='total_bill', kind='box', hue='sex', height=7, aspect=1.5)
```

```python
# single boxplot
sns.catplot(data=tips, y='total_bill', kind='box')
```

#### Violinplot (Boxplot + KDE)

- It's similar to a boxplot, but it also shows the kernel density estimation (KDE) of the data.
- This gives a better understanding of the distribution of the data.

```python
# violinplot
sns.violinplot(data=tips, x='day', y='total_bill', hue='day')
```

### Categorical Estimate Plots

#### Barplot

- Used to show the central tendency of a categorical variable.
- It can show the mean, median, or any other statistical measure.
- It also shows the confidence interval of the estimate.
- By default, it shows the mean and 95% confidence interval.

```python
# plotting barplot
sns.barplot(data=tips, x='sex', y='total_bill')
```

```python
# using catplot
sns.catplot(data=tips, x='sex', y='total_bill', kind='bar', hue='smoker')
```

```python
# estimator
sns.catplot(data=tips, x='sex', y='total_bill', kind='bar', estimator='median')
```

#### Pointplot

- Similar to barplot but uses points and lines instead of bars.

```python
# pointplot
sns.pointplot(data=tips, x='sex', y='total_bill')
```

```python
# using catplot
sns.catplot(data=tips, x='sex', y='total_bill', kind='point', errorbar=('ci', 95))
```

```python
# errorbar
sns.catplot(data=tips, x='sex', y='total_bill', kind='point', errorbar=('ci', 90))

# errorbars can be removed with errorbar=None
```

#### Countplot

- Used to show the count of observations in each categorical bin using bars.
- It's similar to a histogram but for categorical data.

```python
# countplot -> axis level
sns.countplot(data=tips, x='sex')
```

```python
# countplot -> figure level
sns.catplot(data=tips, x='sex', kind='count')
```

```python
tips
```

```python
# hue
sns.catplot(data=tips, x='sex', kind='count', hue='day')
```

## Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
# regplot
# axes level
# hue parameter is not available
sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
# implot
# axes level
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
```

```python
# residplot
sns.residplot(data=tips, x='total_bill', y='tip')
```

## A second way to plot Facet plots -> FacetGrid

```python
tips.sample(1)
```

```python
# initialize FacetGrid
g = sns.FacetGrid(tips, col="sex", row="smoker", margin_titles=True)

# maping scatter plot
g.map(sns.scatterplot, "total_bill", "tip")
```

```python
g = sns.FacetGrid(tips, col="smoker", margin_titles=True)

# maping point plot
g.map(sns.pointplot, 'sex', 'total_bill')
```

```python
tips.head()
```

```python
g = sns.FacetGrid(tips, col="day", col_wrap=2)

g.map(sns.pointplot, 'sex', 'tip')
```

## Pairplot and PairGrid

```python
iris = sns.load_dataset('iris')
iris.sample(1)
```

### Pair Plot

```python
# pair plot
sns.pairplot(iris)
```

```python
# hue
sns.pairplot(iris, hue='species')
```

### PairGrid

- PairGrid is a class that allows you to create a grid of subplots for pairwise relationships in a dataset.

```python
# using PairGrid
g = sns.PairGrid(data=iris, hue='species')
g.map(sns.scatterplot)
```

```python
# map_diag -> map_offdiag
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.kdeplot)
```

## Joint Plot and JointGrid (Multiplots)

### Joint Plot

- shows both a bivariate relationship and the univariate distributions of each variable.

```python
# Joint Plot
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

```python
# hue
sns.jointplot(data=tips, x='total_bill', y='tip', kind='hist', hue='time')
```

### JointGrid

- JointGrid is a class that allows you to create a grid for plotting joint and marginal distributions.

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.scatterplot, sns.histplot)
```

```python
# hist and kdeplot 
g = sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.histplot, sns.kdeplot)
```
