```python
# importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Plot Size:** Plot Size can be adjusted using `plt.figure(figsize=(width, height))` where width and height are in inches.

```python
# customizing plot size
plt.figure(figsize=(10,7))

sales = [500, 490, 495, 430, 590, 643, 749, 801, 920, 940, 1120, 1002]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

plt.bar(months, sales)
```

**Annotation:**

- Refers to adding text or labels to specific points or areas on a plot to provide additional information or context.
- Can help highlight important data points, trends, or features in the visualization.

```python
# sales data
sales = [500, 490, 495, 430, 590, 643, 749, 801, 920, 940, 1120, 1002]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# plotting bar chart for sales data
plt.bar(months, sales)

# using annotation to show exact sales value on top of each bar
for i, v in enumerate(sales):
    plt.text(months[i], sales[i] + 5, v, ha='center')
```

```python
# sample dataframe of students, their study hours and exam scores
students = ["Ali", "Sara", "Rizwan", "Eman", "Ahsan", "Maira", "Sumaira", "Sadia"]

study_hours = [2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [50, 55, 60, 65, 70, 78, 85, 90]

df = pd.DataFrame({
    "student": students,
    "study_hours": study_hours,
    "exam_score": exam_scores
})

# scatter plot for study hours vs exam scores
plt.scatter(df['study_hours'], df['exam_score'])

# using annotation to show student names on each point
for i, v in enumerate(students):
    plt.annotate(v, (df['study_hours'][i] + 0.1, df['exam_score'][i]))
```

**Horizontal and Vertical Lines** 

- We can add horizontal and vertical lines to a plot using `plt.axhline()` and `plt.axvline()` functions respectively.
- These lines can be used to indicate thresholds, reference points, or specific values on the axes.

```python
# loading dataset
df = pd.read_csv('./Datasets/batter.csv').head(25)
df.head()
```

```python
plt.figure(figsize=(15,7))
plt.scatter(df['avg'], df['strike_rate'])

plt.axhline(df['strike_rate'].mean(), color='red', linestyle='--') # horizontal line at mean strike rate
plt.axvline(df['avg'].mean(), color='blue', linestyle='--') # vertical line at mean average

for i, player in enumerate(df['batter']):
    plt.annotate(player, (df['avg'][i], df['strike_rate'][i]))

plt.xlabel('Batting Average')
plt.ylabel('Strike Rate')
plt.title('Batting Average vs Strike Rate with Mean Lines')
plt.show()
```

## Subplots

- Subplots allow us to create multiple plots within a single figure.
- Useful for when we are dealing with multiple variables or want to compare different visualizations side by side.

```python
# loading dataset
df = pd.read_csv('./Datasets/batter.csv')
df.head()
```

```python
# plotting 2 subplots
fig, ax = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(7,7))

ax[0].scatter(df['avg'], df['strike_rate'])
ax[1].scatter(df['avg'], df['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
# plotting 4 subplots
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,10))

ax[0,0].scatter(df['avg'], df['strike_rate']) 
ax[0,1].scatter(df['avg'], df['runs'])
ax[1,0].hist(df['avg'])
ax[1,1].hist(df['runs'])

# setting titles and labels for each subplot
ax[0,0].set_title('Avg Vs Strike Rate')
ax[0,0].set_xlabel('Avg')
ax[0,0].set_ylabel('Strike Rate')

ax[0,1].set_title('Avg Vs Runs')
ax[0,1].set_xlabel('Avg')
ax[0,1].set_ylabel('Runs')

ax[1,0].set_title('Distribution of Avg')
ax[1,0].set_xlabel('Avg')
ax[1,0].set_ylabel('Frequency')

ax[1,1].set_title('Distribution of Runs')
ax[1,1].set_xlabel('Runs')
ax[1,1].set_ylabel('Frequency')

plt.tight_layout() # adjust spacing between subplots
plt.show()
```

## Colored Scatter Plots

```python
# loading 'iris' dataset from seaborn
iris = sns.load_dataset('iris')

iris.head()
```

We want to plot a scatter plot of `sepal_length` vs `petal_length` and color the points based on the species of the iris flower.

```python
# first replacing the species name with numeric 0,1,2
iris['species'] = iris['species'].replace({'setosa':0, 'versicolor':1, 'virginica':2})

iris.sample(5)
```

```python
# ploting colored scatter pot
plt.scatter(iris['sepal_length'], iris['petal_length'], c=iris['species'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')


plt.show()
```

Using cmap and colorbar() and alpha parameter to adjust transparency.

```python
plt.scatter(iris['sepal_length'], iris['petal_length'], c=iris['species'], cmap='jet', alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar() # add a colorbar

plt.show()
```

```python
plt.figure(figsize=(10,7))

plt.scatter(iris['sepal_length'], iris['petal_length'], c=iris['species'], cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.colorbar()
plt.show()
```

## 3d Scatter Plot

- Useful for visualizing relationships between three numerical variables.

```python
df = pd.read_csv('./Datasets/batter.csv').head(200)

# ploting 3d scatter plot
fig = plt.figure(figsize=(10,7))
ax = plt.subplot(projection='3d')

ax.scatter3D(df['avg'], df['runs'], df['strike_rate'], color='green')

ax.set_title('3D Scatter Plot of Runs, Avg, and SR')
ax.set_xlabel('Avg')
ax.set_ylabel('Runs')
ax.set_zlabel('SR')

plt.show()
```

## 3D Surface Plot

- Used to visualize three-dimensional data in a continuous surface.
- Useful in Machine Learning to visualize loss functions or decision boundaries.

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

# 3D Surface Plot
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(X, Y, Z, cmap='viridis')
fig.colorbar(p)
```

## Contour Plot

- Graphical representation of a 3D surface on a 2D plane, where contour lines connect points of equal value.

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

# Contour Plot
fig = plt.figure(figsize=(12,8))
ax = plt.subplot()
p = ax.contour(X, Y, Z, cmap='viridis')
fig.colorbar(p)
```

**Contourf Plot**

- Similar to contour plots but filled with colors between the contour lines.
- Provides a more visually appealing representation of the data.

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

# Contour Plot
fig = plt.figure(figsize=(12,8))
ax = plt.subplot()
p = ax.contourf(X, Y, Z, cmap='viridis')
fig.colorbar(p)
```

## Heatmap

- Used to represent data in a matrix format where individual values are represented as colors.
- Useful for visualizing correlations, patterns, or distributions in large datasets.

```python
# dataset
df = pd.read_csv('./Datasets/IPL_Ball_by_Ball_2008_2022.csv')
```

```python
# filtering sixes in first six balls
temp_df = df[(df['ballnumber'].isin([1,2,3,4,5,6])) & (df['batsman_run'] == 6)]
temp_df.head()
```

```python
# creating pivot table for heatmap
grid = temp_df.pivot_table(index='overs', columns='ballnumber', values='batsman_run', aggfunc='count')
grid
```

```python
# plotting heatmap
plt.figure(figsize=(20,10))
plt.imshow(grid)

plt.yticks(df['overs'].unique(), list(range(1,21))) # setting y-ticks as overs 1-20
plt.xticks([0,1,2,3,4,5], list(range(1,7))) # setting x-ticks as ball numbers 1-6
plt.colorbar()

plt.title('Heatmap of Sixes on Each Ball of Each Over in IPL')
plt.xlabel('Ball Number')
plt.ylabel('Over Number')
plt.show()
```
